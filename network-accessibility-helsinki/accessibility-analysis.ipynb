```python
# Specify the path to the shapefile
shapefile_path_districts = r'housing_district16nrm_main.shp'

# Load the shapefile into a GeoDataFrame
gdf = gpd.read_file(shapefile_path_districts)

# Display the first few rows of the GeoDataFrame
gdf.head()

# Specify the path to the shapefile
shapefile_path_network = r'Final_Main_RoadHK2016.shp'

# Load the shapefile into a GeoDataFrame
gdf2 = gpd.read_file(shapefile_path_network)

"""# BELOW CODE WILL VISUALIZE THE DATA IN A SIMPLE WAY 
gdf2.head()
gdf.plot()
gdf2.plot()
plt.show()"""

```

## Visualize the network and label the district index numbers in Helsinki

```python
# Load the network shapefile into a GeoDataFrame


# Create the plot
fig, ax = plt.subplots(figsize=(20, 20))

# Plot the entire district shapefile (all polygons)
gdf2.plot(ax=ax, color='gray', edgecolor='black', alpha=0.5)


# Plot the network shapefile (roads) on top of the polygons
gdf.plot(ax=ax, color='lightgray', linewidth=0.7)

# Set plot title and labels
ax.set_title("District Polygon and Network Components", fontsize=14)


gdf.plot(ax=ax, color='lightgray', edgecolor='black', alpha=0.5)

# Loop through each district and plot its index on the map
for idx, row in gdf.iterrows():
    # Get the centroid of each polygon for placing the index label
    centroid = row['geometry'].centroid
    x, y = centroid.x, centroid.y  # Extract x, y coordinates of the centroid
    
    # Annotate the district index number at the centroid
    ax.text(x, y, str(idx), fontsize=10, ha='center', color='black', weight='bold')



# Display the plot
plt.show()
```

# DISTRICT 19 (32 nodes and 31 edges)

```python
# Load the adjacency matrix from the text file
# Load the adjacency matrix from the text file
adj_matrix = np.loadtxt('DISTRICT19_adjacency_matrix.txt', delimiter="\t", dtype=int)
nodecoord19 = np.loadtxt('node_coordinates_District19.txt', delimiter="\t", dtype=int)
node_dict19 = {i: tuple(coord) for i, coord in enumerate(nodecoord19)}

# Create a graph from the adjacency matrix using networkx (use from_numpy_array)
G_19 = nx.from_numpy_array(adj_matrix)

# Display basic graph information using networkx methods
print("Graph Summary:")
print(f"Number of nodes: {G_19.number_of_nodes()}")
print(f"Number of edges: {G_19.number_of_edges()}")
```

```python
# Calculate betweenness centrality (example)
betweenness_subgraph = nx.betweenness_centrality(G_19)
```

```python

# Plot the subgraph
plt.figure(figsize=(10, 10))
node_dict = {i: tuple(coord) for i, coord in enumerate(nodecoord19)}
# Draw the nodes with size proportional to betweenness centrality on the node coordinates
node_sizes = [betweenness_subgraph.get(node, 0) * 1000 for node in G_19.nodes()]  # Scale betweenness for visualization
nx.draw(G_19, pos=node_dict19, node_size=node_sizes, node_color='red', alpha=0.7)
# Set the title of the plot
plt.title("Subgraph of Intersecting Network - Betweenness Centrality")

# Display the plot
plt.axis('off')  # Turn off axis
plt.show()
```
