# Procedures

```python
# Procedure to plot histogram
def plot_histogram(G):
# Compute the degree of each node
    degrees = [degree for node, degree in G.degree()]
# Plot the degree distribution using a histogram
    plt.hist(degrees, bins=range(min(degrees), max(degrees)+1),rwidth=0.8)
    plt.xlabel('Degree')
    plt.ylabel('Frequency')
    plt.show()

# Procedure to generate dynamic random graph
def dynamic_random_graph(number_of_links,time_steps):
#Define an empty graph
    dynamic_graph = nx.Graph() 
#Loop through all time_steps
    for node in range(time_steps):
#Add the new node
        dynamic_graph.add_node(node)
#If there are sufficient old nodes to connect to
        if node>=number_of_links:
#Sample nodes to connect new node to
            sample = random.sample(list(range(node)),number_of_links)
#Connect new node to nodes in sample
            dynamic_graph.add_edges_from([(node,old_node) for old_node in sample])
    return dynamic_graph
```

# Python exercise 1: Static random graph  

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

def generate_random_graph(n, p):
    """Generates an Erdős–Rényi random graph."""
    return nx.erdos_renyi_graph(n, p)

def plot_graph(graph, p):
    """Plots the given graph."""
    plt.figure(figsize=(6, 6))
    nx.draw(graph, with_labels=True, node_size=300, font_size=8)
    plt.title(f"Erdős–Rényi Graph (p={p})")
    plt.show()

def plot_degree_distribution(graph, p):
    """Plots the degree distribution of the graph."""
    degrees = [d for _, d in graph.degree()]
    plt.figure(figsize=(6, 4))
    plt.hist(degrees, bins=np.arange(min(degrees), max(degrees) + 1, 1), alpha=0.75, edgecolor='black')
    plt.xlabel("Degree")
    plt.ylabel("Frequency")
    plt.title(f"Degree Distribution (p={p})")
    plt.show()

def check_connectivity(graph, p):
    """Checks if the graph is connected and prints the result."""
    is_connected = nx.is_connected(graph)
    print(f"Graph with p={p} is {'connected' if is_connected else 'not connected'}.")

# Parameters
n = 100
p_values = [0.2, 0.5, 0.8]

# Generate and analyze graphs
for p in p_values:
    G = generate_random_graph(n, p)
    check_connectivity(G, p)
    plot_graph(G, p)
    plot_degree_distribution(G, p)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Answer 3a: Generate static random graph with 100 nodes and probability 0.075
static_graph = nx.erdos_renyi_graph(100, 0.075)
# Answer 3b: Check if graph is connected
is_connected_static = nx.is_connected(static_graph)
print(f"Static Graph is {'connected' if is_connected_static else 'not connected'}.")
# Answer 3c: Calculate number of edges
num_edges_static = static_graph.number_of_edges()
print(f"Number of edges in static graph: {num_edges_static}")

# Answer 3d: Generate dynamic random graph with 100 nodes and approximately same number of edges
dynamic_graph = nx.barabasi_albert_graph(100, 3)
# Answer 3e: Calculate number of edges
num_edges_dynamic = dynamic_graph.number_of_edges()
print(f"Number of edges in dynamic graph: {num_edges_dynamic}")

# Answer 3f: Generate Watts-Strogatz graph with 100 nodes and approximately same number of edges
watts_graph = nx.watts_strogatz_graph(100, 8, 0.2)
# Answer 3g: Calculate number of edges
num_edges_watts = watts_graph.number_of_edges()
print(f"Number of edges in Watts-Strogatz graph: {num_edges_watts}")

# Answer 3h: Generate Barabási-Albert graph with 100 nodes and approximately same number of edges
barabasi_graph = nx.barabasi_albert_graph(100, 4)
# Answer 3i: Calculate number of edges
num_edges_barabasi = barabasi_graph.number_of_edges()
print(f"Number of edges in Barabási-Albert graph: {num_edges_barabasi}")

# Answer 3j: Calculate average clustering coefficient for all 4 graphs
names = {static_graph: 'static random',
         dynamic_graph: 'dynamic random',
         watts_graph: 'watts-strogatz',
         barabasi_graph: 'barabasi-albert'}

print('Average clustering coefficient:')
for graph in [static_graph, dynamic_graph, watts_graph, barabasi_graph]:
    cluster_coefficient = nx.average_clustering(graph)
    print(names[graph], ':', cluster_coefficient)

# Answer 4a: Calculate number of local bridges
print('Number of local bridges:')
def count_local_bridges(graph):
    local_bridges = 0
    for edge in graph.edges():
        graph.remove_edge(*edge)
        if nx.has_path(graph, edge[0], edge[1]) == False:
            local_bridges += 1
        graph.add_edge(*edge)
    return local_bridges

for graph in [static_graph, dynamic_graph, watts_graph, barabasi_graph]:
    number_of_bridges = count_local_bridges(graph)
    print(names[graph], ':', number_of_bridges)

# Answer 4b: Average span
print('Average span:')
def average_span(graph):
    span_values = []
    for edge in graph.edges():
        graph.remove_edge(*edge)
        if nx.has_path(graph, edge[0], edge[1]):
            span_values.append(nx.shortest_path_length(graph, edge[0], edge[1]))
        graph.add_edge(*edge)
    return np.mean(span_values) if span_values else 0

for graph in [static_graph, dynamic_graph, watts_graph, barabasi_graph]:
    av_span = average_span(graph)
    print(names[graph], ':', av_span)
```
