```python
import networkx as nx
import matplotlib.pyplot as plt

# Define the graph correctly
G = nx.Graph()
edges = [(1,2), (1,3), (2,3), (2,4), (3,4), (4,5), (4,6), (5,6)]
G.add_edges_from(edges)

# Draw the graph
nx.draw(G, with_labels=True, node_color='lightblue', edge_color='gray')

# Show the plot
plt.show()

cycles = nx.cycle_basis(G)

# Print the cycles found
print("Cycles in the graph:", cycles)
```

```python
#Answer 1c: Calculate the cycle basis:
#minimal collection of cycles such that any cycle in the network can be written as a sum of cycles in the basis
```

```python
#Python exercise 2: Subgraph
#Answer 2a: Vertex induced subgraph
nodes_subset = {1,2,3,4}  # Nodes to keep
G_vertex_subgraph = G.subgraph(nodes_subset)

# Draw vertex-induced subgraph
plt.figure(figsize=(6, 4))
nx.draw(G_vertex_subgraph, with_labels=True, node_color='lightblue', edge_color='gray')
plt.title("Vertex-Induced Subgraph")
plt.show()
```

```python
#Answer 2b: Edge-induced subgraph
G_edge_subgraph = nx.Graph()
edges_subset = [(1,2), (2,3), (3,4), (2,4)]  # Edges to keep
G_edge_subgraph.add_edges_from(edges_subset)

# Draw edge-induced subgraph
plt.figure(figsize=(6, 4))
nx.draw(G_edge_subgraph, with_labels=True, node_color='lightgreen', edge_color='gray')
plt.title("Edge-Induced Subgraph")
plt.show()
```

```python
#Answer 3a: Strongly connected component
import networkx as nx
import matplotlib.pyplot as plt

# Maak een gerichte graaf (digraph)
H = nx.DiGraph()

# Voeg de randen toe zoals in de afbeelding
edges = [(1,2), (2,3), (3,4), (4,8), (2,6), (6,7), (7,8), (1,5), (5,6), (6,5)]
H.add_edges_from(edges)

# Vind sterk verbonden componenten
sccs = list(nx.strongly_connected_components(H))

# Filter niet-triviale SCC’s (meer dan 1 knoop)
non_trivial_sccs = [scc for scc in sccs if len(scc) > 1]

# Print de niet-triviale SCC’s
print("Niet-triviale sterk verbonden componenten vóór de wijziging:", non_trivial_sccs)

# Visualiseer de graaf
plt.figure(figsize=(6,4))
nx.draw(H, with_labels=True, node_color="lightblue", edge_color="gray", arrows=True)
plt.title("Oorspronkelijke Graaf H")
plt.show()
```

```python
#Answer 3b: Add edge (3,2)
H.add_edge(3,2)

# Vind de nieuwe SCC's
new_sccs = list(nx.strongly_connected_components(H))
new_non_trivial_sccs = [scc for scc in new_sccs if len(scc) > 1]

# Print de niet-triviale SCC’s na toevoegen van de rand
print("Niet-triviale sterk verbonden componenten na toevoeging van (3,2):", new_non_trivial_sccs)

# Visualiseer de aangepaste graaf
plt.figure(figsize=(6,4))
nx.draw(H, with_labels=True, node_color="lightgreen", edge_color="gray", arrows=True)
plt.title("Graaf H na toevoeging van (3,2)")
plt.show()
#Answer 3c: Strongly connected component
```
