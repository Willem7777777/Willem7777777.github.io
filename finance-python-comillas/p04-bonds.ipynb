## 5. **Exercise:** Convexity

Following this nomenclature:
- $P$: price.
- $y$: yield to maturity.
- $N$: number of periods.
- $t_i$: time in years of period $i$ considering $k$ periods per year.
- $PV_i$: present value of $CF_i$ at time $t_i$.


**Convexity** can be calculated this way:

$$
\texttt{convexity} = \frac{1}{p} \cdot \frac{\partial^2 P}{\partial y^2}
$$

$$
\frac{\partial^2 P}{\partial y^2} = \frac{1}{\left( 1 + \frac{y}{k}\right)^2} \cdot \sum_{i=1}^N t_i \cdot \left( t_i+\frac{1}{k} \right) \cdot PV_i
$$

Write a function `bond_convexity()` that returns the convexity of a given bond (about 4 lines of code).  
**Follow exactly the equation above!** Use the column `convexity_term` to store the inner term of the summation.

````python


 def bond_convexity(data, price, yield_to_maturity, coupons_per_year):
    periods_per_year = coupons_per_year
    yk = yield_to_maturity / periods_per_year
    N = len(data['PV'])
    
    # Calculate the inner term of the summation
    convexity_term = sum((data['time'][i] * (data['time'][i] + 1) / periods_per_year) * data['PV'][i] for i in range(N))
    
    # Calculate convexity
    convexity = (1 / price) * (1 / ((1 + yk) ** 2)) * convexity_term
    
    # Store the convexity term in the data dictionary
    data['convexity_term'] = convexity_term
    
    return convexity, data
    
  
````

## 6. **Run:** Plot price, duration and covexity
Assuming your code is right, this should work like a charm...

Price at the point (i.e., yield) where the function is approximated.

$$P_0 = price(ytm_0)$$

First degree: using **modified duration** $D^*$ alone.

$$first(ytm) = P_0 - \left(ytm - ytm_0 \right) \cdot P_0 \cdot D^{*}$$

Second degree: using both **modified duration** $D^*$ and **modified convexity** $C^*$.

$$second(ytm) = P_0 - \left(ytm - ytm_0 \right) \cdot P_0 \cdot D^{*} + \frac{1}{2} \cdot \left(ytm - ytm_0 \right)^2 \cdot P_0 \cdot C^{*} $$

The function `plot_duration_convexity(...)` displays the price as a function of the YTM along with the first- and second-degree approximations.
- `data`: bond data.
- `number_coupons_per_year`
- `modified_duration`
- `convexity`
- `plot_price` and `plot_yield_to_maturity`: where to apply the approximation.

**IMPORTANT:** the `modified_duration` and `convexity` provided as arguments **must** be calculated at the given `plot_price` and `plot_yield_to_maturity`. Otherwise weird things will happen...

````python
def plot_duration_convexity(data,
                            price,
                            yield_to_maturity, 
                            coupons_per_year,
                            modified_duration,
                            convexity):
    
    """ plot of price as function of YTM, along with first- and second-degree approximations """

    # helper function to plot price
    def _price(ytm):
        price, _ = bond_value(data, ytm, coupons_per_year)
        return price
    
    # helper function to plot first-degree (linear) approximation
    def _duration(ytm):
        return price - (ytm - yield_to_maturity) * price * modified_duration

    # helper function to plot second-degree (quadratic) approximation
    def _duration_convexity(ytm):
        return _duration(ytm) + 0.5 * price * convexity * (ytm - yield_to_maturity)**2

    # prepare a neat table
    yield_curve = pd.DataFrame()
    yield_curve['yield']                       = np.linspace(0, 2*yield_to_maturity, 100)
    yield_curve['price']                       = yield_curve['yield'].apply(_price)
    yield_curve['aprox. duration']             = yield_curve['yield'].apply(_duration)
    yield_curve['aprox. duration & convexity'] = yield_curve['yield'].apply(_duration_convexity)

    # plot
    melted_df = yield_curve.melt(id_vars='yield')
    fig = px.line(melted_df, x='yield', y='value', color='variable', line_dash='variable', width=600, height=500)
    
    # vertical line at price
    fig.add_hline(y=price, line_color='black', line_dash='dot', line_width=1)
    
    # vertical line at YTM
    fig.add_vline(x=yield_to_maturity, line_color='black', line_dash='dot', line_width=1)
    
    # title, axis labels and use percentages in axis
    fig.update_layout(
        width=1200,
        height=600,
        xaxis= { 'tickformat': '.1%'},
        title='Real price & Duration and Convexity aproximations',
        xaxis_title='YTM [%]',
        yaxis_title='price [$]',
    )

    fig.show()
````

## 7. **Run:** Making your life easier...

As you may have noticed already, all these sequential steps can be grouped together in a single function that calculates all bond features in one go.  
Let's name it `bond_report()` and it is given to you below:

````python
def bond_report(face_value,
                maturity,
                yield_to_maturity,
                coupons_per_year,
                coupon_rate):

    # Populate cashflows
    data = bond_data(
        face_value,
        maturity,
        coupons_per_year,
        coupon_rate)
    
    # Calculate price
    price, data = bond_value(
        data,
        yield_to_maturity,
        coupons_per_year)
    
    # bond yield to maturity
    # (a single value, even if discount rates change for each cashflow)
    bond_ytm, data = bond_yield_to_maturity(
        data,
        price,
        coupons_per_year)
    
    # Calculate duration
    macaulay, modified_duration, data =  bond_duration(
        data,
        price,
        bond_ytm,
        coupons_per_year)
    
    # Calculate convexity
    convexity, data =  bond_convexity(
        data,
        price,
        bond_ytm,
        coupons_per_year)
    
    # print some key data
    print('BOND REPORT (Summary of key features)')
    print('---------------------------------------------')
    print(f'       face_value  = {face_value:.3f}$')
    print(f'         maturity  = {maturity} years')
    print(f'      coupon rate  = {coupon_rate:.3%}')
    print(f'     coupons/year  = {coupons_per_year}')
    print(f'           coupon  = {face_value*coupon_rate/coupons_per_year:.3f}$')
    print('---------------------------------------------')
    
    print(f'             price = {price:.3f}$')
    print(f' yield to maturity = {yield_to_maturity:.3%} (calculated)')
    print(f' Macaulay duration = {macaulay:.3f} years')
    print(f' modified duration = {modified_duration:.3f} [Δ%/1%]')
    print(f'         convexity = {convexity:.3f}')
    print('---------------------------------------------')

    # plot
    plot_duration_convexity(data,
                            price,
                            yield_to_maturity, 
                            coupons_per_year,
                            modified_duration,
                            convexity)

    # return table with all calculations
    return data
````

## 10. **Exercise:** What about TSIR?

The current implementation of `bond_value()` uses a constant *yield to maturity* for every period. Let's create an alternate version of the valuation function that supports different discount rates per period.  

Now:
$$
P = \sum_{i=1}^N \frac{CF_i}{\left( 1 + \frac{y}{k} \right)^i}
$$

Proposed changes:

$$
P = \sum_{i=1}^N \frac{CF_i}{\left( 1 + y_i \right)^i}
$$

Write a function `bond_value_tsir()` that values a bond taking into account different discount rates per cashflow (about 4 lines of code).  
For simplity, just assume that the length of `discount_rates` is correct.

````python
def bond_value_tsir(data,
                    discount_rates,
                    coupons_per_year):

  
    # Calculate the total bond value using different discount rates per cash flow
    bond_value = sum(data[f'C{i}'] / ((1 + discount_rates[i - 1]) ** (i / coupons_per_year)) for i in range(1, coupons_per_year + 1))
    
    return bond_value
    
````

## 11. Exercise

>The **U.S. Goverment** issues a bond with a coupon rate of **6%** (it makes coupon payments **twice a year**) with **4 years** to maturity. After interpolating the TSIR curve, the discount rates every six months are:

```
DISCOUNT_RATES = [0.00, 0.03, 0.04, 0.045, 0.05, 0.0525, 0.055, 0.0575, 0.06]
```

- **Calculate the price, YTM, Macaulay duration, modified duration, and convexity.**
- **How much does the price change \[%\] if interest rates go up 1%? Plot the price, the first- and second-degree approximations.**


**Hint:** Basically, follow the steps in `bond_report()` with slight differences...

````python
import numpy as np
import matplotlib.pyplot as plt

# Given parameters
coupon_rate = 0.06  # Coupon rate
face_value = 1000  # Face value
years_to_maturity = 4  # Years to maturity
coupons_per_year = 2  # Coupon payments per year
total_periods = years_to_maturity * coupons_per_year  # Total periods

# Given discount rates
DISCOUNT_RATES = [0.00, 0.03, 0.04, 0.045, 0.05, 0.0525, 0.055, 0.0575, 0.06]

# Step 1: Calculate cash flows
cash_flows = [face_value * coupon_rate / coupons_per_year] * total_periods
cash_flows[-1] += face_value  # Last cash flow is the face value

# Step 2: Calculate YTM
ytm = DISCOUNT_RATES[-1]

# Step 3: Calculate bond value with TSIR
bond_value = sum(cash_flows[i] / ((1 + DISCOUNT_RATES[i]) ** (i / coupons_per_year)) for i in range(total_periods))

# Step 4: Calculate Macaulay duration
macaulay_duration = sum((i + 1) * cash_flows[i] / bond_value for i in range(total_periods))

# Step 5: Calculate modified duration
modified_duration = macaulay_duration / (1 + ytm / coupons_per_year)

# Step 6: Calculate convexity
convexity_term = sum(((i + 1) * (i + 2) * cash_flows[i]) / ((1 + DISCOUNT_RATES[i]) ** (i / coupons_per_year + 2)) for i in range(total_periods))
convexity = convexity_term / bond_value

# Step 7: Calculate bond value if interest rates increase by 1%
new_ytm = ytm + 0.01
new_bond_value = sum(cash_flows[i] / ((1 + new_ytm / coupons_per_year) ** (i / coupons_per_year)) for i in range(total_periods))

# Calculate the price change in percentage
price_change_percent = ((new_bond_value - bond_value) / bond_value) * 100

# Step 8: Plot bond price and approximations
x = np.arange(0, years_to_maturity + 0.1, 0.1)
bond_price = np.array([bond_value] * len(x))
first_degree_approximation = bond_value - modified_duration * bond_value * (new_ytm - ytm) * (x * coupons_per_year)
second_degree_approximation = bond_value - (modified_duration * bond_value * (new_ytm - ytm) * (x * coupons_per_year)) + (0.5 * convexity * bond_value * (new_ytm - ytm) ** 2 * (x * coupons_per_year) ** 2)

# Plot
plt.plot(x, bond_price, label='Bond Price')
plt.plot(x, first_degree_approximation, label='1st-degree Approximation')
plt.plot(x, second_degree_approximation, label='2nd-degree Approximation')
plt.xlabel('Years to Maturity')
plt.ylabel('Price')
plt.title('Bond Price and Approximations')
plt.legend()
plt.show()

# Print the results
print("Price: $", round(bond_value, 2))
print("YTM:", round(ytm * 100, 2), "%")
print("Macaulay Duration:", round(macaulay_duration, 2), "years")
print("Modified Duration:", round(modified_duration, 2), "years")
print("Convexity:", round(convexity, 2))
print("Price Change if rates go up by 1%:", round(price_change_percent, 2), "%")
````
