## 2. **Exercise:** Write a Payback function

How long does it take for the project to "pay back" its initial investment? The **Payback period** is the number of year to recover the initial costs.

$$
\sum_{t=1}^{payback} CF_t = I_0
$$

Write a function that calculates the payback of a stream of `cashflows` (about 7 lines of code).  
To make things simpler, do not interpolate, just return the integer number of years required to pay back.  
If the initial investment is never paid back, return `None`.

**IMPORTANT:** The initial investment $I_0$ is provided as the first item of the `cashflows` list (with negative sign).

```python
#De formule "∑𝑡=1𝑝𝑎𝑦𝑏𝑎𝑐𝑘𝐶𝐹𝑡=𝐼0" geeft aan dat we de som van alle kasstromen 
#(CFt) van tijdstap 1 (t=1) tot het laatste jaar of de laatste periode (t=p) 
#moeten berekenen. Als we deze som uitvoeren, krijgen we de totale kasstroom 
#gedurende de levensduur van het project of de investering
#Het idee is om te zien hoe lang het duurt voordat de totale kasstroom de initiële investering 
#(I0) terugverdient. Als de som van de kasstromen groter of gelijk is aan de initiële investering, 
#betekent dit dat de investering op dat moment is terugverdiend. 
#De waarde van t op dat moment is de terugverdientijd.

#initial investment is de -1000, dus hoe lang duurt het voordat de rest bij elkaar 
#1000 is

def payback_period(cashflows):
    initial_investment=cashflows[0]  #plek 1 van lijst 
    cumulative_cashflow=initial_investment 
    years=0
    
    for cashflow in cashflows[1:]:       #van plek 1 tot einde lijst
        cumulative_cashflow+=cashflow
        years+=1
        if cumulative_cashflow>=0:
            return years
        
    print("the payback time is ",years)
        
        
        
        
TEST_CASHFLOWS = [-1000, 100, 200, 300, 400, 500]
payback_period(TEST_CASHFLOWS)   
    
    
```

## 3. **Exercise:** Write an NPV function

The **Net Present Value (NPV)** is the total Present Value (PV) of a project's future cashflows less the initial investment (here $CF_0<0$).

$$
NPV = CF_0 + \sum_{t=1}^{n}\frac{CF_t}{(1+r)^t}
$$

Write a function that calculates the NPV of a stream of `cashflows` at a given `discount_rate` (about 5 lines of code).  
Do not use `npv()` from package `numpy-financial`, **code it yourself!**

**IMPORTANT:** The initial investment $I_0$ is provided as the first item of the `cashflows` list (with negative sign).

```python
#De NPV berekent de huidige waarde van al die toekomstige kasstromen en vergelijkt 
#deze met de initiële investering. Het resultaat kan positief, negatief of nul zijn en 
#biedt belangrijke inzichten in de vraag of een investering winstgevend is.



#Positieve NPV: Als de NPV positief is, betekent dit dat de verwachte toekomstige kasstromen de initiële investering overtreffen. Dit duidt meestal op een winstgevende investering, en het project wordt als levensvatbaar beschouwd.
#Negatieve NPV: Als de NPV negatief is, betekent dit dat de verwachte toekomstige kasstromen onvoldoende zijn om de initiële investering te dekken. In dit geval wordt het project als niet-levensvatbaar beschouwd, omdat het naar verwachting verliesgevend zal zijn.


def npv(discount_rate, cashflows):
    npv = cashflows[0]  #op plek 1 in lijst = initiele investering
    
    for t in range(1, len(cashflows)):  #je gaat van jaartal 1 naar lengte cashflow - 1= 5 -> lengte gaat niet indexen 
        npv+=cashflows[t]/(1 + discount_rate) ** t
    return npv
        

TEST_CASHFLOWS = [-1000, 100, 200, 300, 400, 500]  #bij 100 is het 1 
TEST_RATE = 0.05
result = npv(TEST_RATE, TEST_CASHFLOWS)
print("The net present value is:", result)
    

```

## 4. **Exercise:** Write an IRR function

The **Internal Rate of Return (IRR)** is the discount rate that sets NPV to zero.

$$
CF_0 + \sum_{t=1}^{n}\frac{CF_t}{(1+IRR)^t} = 0
$$

Write a function that calculates the IRR of a stream of `cashflows`.  
As this is a complex calculation, this time you are allowed to call the `irr()` function from package `numpy-financial`. **Look it up in the documentation!**  
So, essentially, your function is a wrapper around it (about 1 line of code).

**IMPORTANT:** The initial investment $I_0$ is provided as the first item of the `cashflows` (with negative sign).

```python
#DE IRR 
#De IRR is de disconteringsvoet of het rentepercentage dat de NPV van een reeks 
#kasstromen op nul zet. Met andere woorden, het is de rentevoet waarmee de huidige 
#waarde van de verwachte kasstromen gelijk wordt aan de initiële investering. 
#Het is het punt waarop de investering 'break-even' wordt, wat betekent dat de 
#verwachte opbrengsten precies overeenkomen met de initiële investeringskosten.
#geen wist geen verlies = break even 
#numpy-financial: for the IRR

def irr(cashflows):
    return npf.irr(cashflows) #hier stop je de hele cashflow in-> dan weet je bij welke irr break even point 
    
TEST_CASHFLOWS_IRR = [-1256.64, 100, 200, 300, 400, 500]#initial_investment=cashflows[0]
irr(TEST_CASHFLOWS_IRR)   
```

## 5. **Exercise:** Calculate crossover rate of two projects

The **Crossover rate** or **Fisher intersection** is the discount rate at which both projects have the same NPV.

$$NPV_A = NPV_B$$

To find the crossover rate:  
- Calculate the difference between the cash flows of the two projects for each period.
- Find the discount rate (IRR) for which the NPV of these differential cash flows is zero.

Write a function that calculates the crossover rate (about 2 lines of code).  
This new function must call your `irr(...)` function.

**IMPORTANT:** From now on, let's use pandas `Series` instead of regular Python lists.  

To make things simpler, **assume both projects have the same number of cashflows.**

```python
#fisher of crossover rate is de disconteringsvoet (IRR) waarbij de Netto Contante 
#Waarde (NPV) van twee verschillende investeringsprojecten of kasstroomreeksen gelijk 
#is aan nul. Met andere woorden, het is de rentevoet waarbij beide projecten dezelfde 
#waarde hebben in termen van de huidige waarde van hun toekomstige kasstromen.

##De NPV berekent de huidige waarde van al die toekomstige kasstromen en vergelijkt 
#deze met de initiële investering. Het resultaat kan positief, negatief of nul zijn en 
#biedt belangrijke inzichten in de vraag of een investering winstgevend is.

#omdat het wordt gebruikt om te bepalen welk project of welke investering aantrekkelijker
#is als er meerdere opties zijn. Als de disconteringsvoet (IRR) hoger is dan de 
#crossover rate -> dan npv niet gelijk aan elkaar 
#dan heeft het ene project een hogere NPV en wordt het beschouwd als 
#de betere keuze. Als de IRR lager is 
#dan de crossover rate, heeft het andere project een hogere NPV en is het de betere 
#keuze

#om de interne rentabiliteit (IRR) te berekenen voor de differentiële kasstromen. 
#Dit is het punt waarop de NPV van het ene project gelijk is aan de NPV van het andere
#project, omdat de differentiële kasstromen nul zijn. 
#De functie retourneert de berekende IRR als resultaat

#Pandas maakt het gemakkelijk om gegevens uit te lijnen op basis van labels. 
#In dit geval kunnen we de kasstromen van beide projecten eenvoudig van elkaar aftrekken
#om de differentiële kasstromen te berekenen

#dus je doet ene cashflow min andere cashflow = 0 -> daar de irr van

def crossover_rate(cashflows_a, cashflows_b):
    differential_cashflows=cashflows_a - cashflows_b #het verschil in kasstromen -> wordt gestopt in npf formule 
    return npf.irr(differential_cashflows)   #irr -> differential cashflow iszero op dit punt
    
TEST_CASHFLOWS_1 = pd.Series([-10000, 10000, 1000, 1000])
TEST_CASHFLOWS_2 = pd.Series([-10000, 1000, 1000, 12000])
crossover_rate(TEST_CASHFLOWS_1, TEST_CASHFLOWS_2)   
```

```python
def plot_crossover_rate(cashflows_1, cashflows_2, min_r, max_r, steps):

    df = pd.DataFrame()

    # Calculate the table and the points to plot
    discount_rates = np.linspace(min_r, max_r, steps)
    df['discount_rates'] = discount_rates

    # Option 1: list comprehension
    df['npv_a'] = [npv(rate, cashflows_1) for rate in discount_rates]
    df['npv_b'] = [npv(rate, cashflows_2) for rate in discount_rates]

    # Option 2: lambda function
    #df['npv_a'] = df['discount_rates'].apply(lambda r: npv(r, cashflows_1))
    #df['npv_b'] = df['discount_rates'].apply(lambda r: npv(r, cashflows_2))
                                             
    # Option 3: pandas apply()
    #df['npv_a'] = df['discount_rates'].apply(npv, args=(cashflows_1,))
    #df['npv_b'] = df['discount_rates'].apply(npv, args=(cashflows_2,))
    
    # Create the interactive plot
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(x=df['discount_rates'], y=df['npv_a'], mode='lines', name='Project A'))
    fig.add_trace(go.Scatter(x=df['discount_rates'], y=df['npv_b'], mode='lines', name='Project B'))
    
    fig.update_layout(title='Payoff Profiles of Two Projects',
                      xaxis_title='Discount Rate',
                      width=800,
                      height=600,
                      yaxis_title='Net Present Value (NPV)',
                      xaxis=dict(tickformat=".2%"),
                      legend_title="Projects")
    
    fig.show()
    
    return df

#het ligt aan de discount rate welk project winstgevender is 
#winstgevendheid wordt gebaseerd op de NPV -> NPV hoog winstgevend 
#
```
