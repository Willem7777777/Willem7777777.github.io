```python
#Importeer de module networkx en kort af als nx
import networkx as nx
#Importeer de module matplotlib.pyplot en kort af als plt
import matplotlib.pyplot as plt

#Definieer een lege graaf K
K = nx.DiGraph()
#Definieer de eerste groep knopen (de studenten)
studenten = ['A','B','C','D','E','F','G']
#Voeg de knopen uit de eerste groep toe (de studenten)
K.add_nodes_from(studenten, bipartite=0) 
#Definieer de tweede groep knopen (de klussen)
#1 = stofzuigen, 2 = koelkast, 3 = wcs, 4 = badkamer, 5 = keuken, 6 = stoffen, 7 = dweilen
klussen = [1,2,3,4,5,6,7]
#Voeg de knopen uit de tweede groep toe (de klussen)
K.add_nodes_from(klussen, bipartite=1)
#Voeg de takken toe
takken = [('A',1),('A',4),('B',2),('B',3),('B',7),('C',4),('C',5),('D',1),('D',2),('D',6),('E',1),('E',4), ('E',6),('F',2),('F',3),('F',7),('G',3),('G',4),('G',5)]
K.add_edges_from(takken)

#Definieer een opzoektabel (dictionary) voor de coordinaten van de knopen
#lijst.index(i) geeft de index van element i  in de lijst
coord_studenten = {i: [1,studenten.index(i)] for i in studenten}
coord_klussen = {i: [2,klussen.index(i)] for i in klussen}

#Combineer de twee opzoektabellen tot 1 tabel
coord = {**coord_studenten,**coord_klussen}

#Teken de graaf
nx.draw_networkx(K, pos=coord)
plt.show()
```

```python
#Kopieer de graaf K
KC=K.copy()
#Definieer takken vanaf nieuwe knoop 0 naar alle studenten
takken0 = [(0,i) for i in studenten]
#Definieer takken naar nieuwe knoop 8 vanaf alle klussen 
takken8=[(8,i) for i in klussen]
#Voeg takken toe aan graaf
KC.add_edges_from(takken0+takken8)
#Geef alle takken capaciteit 1
for i,j in KC.edges():
    KC[i][j]['capacity']=1
    
#Bereken de maximale stroom door de graaf van knoop 0 naar knoop 8
totaal,koppeling=nx.maximum_flow(KC,0,8)
print(totaal)
```

```python
#Print de student,klus koppelingen
for i in studenten:
    for j in klussen:
        if (i,j) in KC.edges() and koppeling[i][j]==1:
            print('student', i,'gaat klus', j, 'doen')
```

Is er met deze nieuwe beperkingen nog steeds een goed oplossing te vinden?

```python
#Definieer eerst de takken die nu niet meer toegestaan zijn
niettoegestaan=[('D',1),('D',6),('G', 3),('A',4)]
#Maak een kopie van originele graaf
K2=K.copy()
#Verwijder de takken uit deze nieuwe graaf
K2.remove_edges_from(niettoegestaan)
#Bepaal de maximale toewijzing
koppeling2=nx.bipartite.maximum_matching(K2, top_nodes=studenten)
#Print de student,klus koppelingen
for i in studenten: 
    print('student', i , 'gaat klus', koppeling2[i], 'doen')
```
