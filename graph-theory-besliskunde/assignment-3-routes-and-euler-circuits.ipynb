In de cel hieronder wordt opnieuw de graaf G gedefinieerd met alle mogelijke verbindingen tussen de kamers en de locaties. Tevens is voor iedere verbinding de afstand opgenomen als gewicht bij de takken.   
Definieer deze graaf door onderstaande cel te runnen.

```python
#Importeer de benodigde modules
import networkx as nx
from matplotlib import pyplot as plt

#Definieer de takken van de graaf met hun lengte als gewicht
edges = [(0, 16,900),(0, 18,630),(1, 9,1630), (1, 21,500), (2, 9,300), (2, 13,800), (2, 22,590), (3, 5,1110), (3, 11,580), (3, 13,920), (4, 5,380), (4, 6,670), (4, 18,750), (6, 12,640), (6, 14,990), (6, 15,600), (7, 11,100), (7, 12,110), (7, 13,260), (7, 14,410), (8, 19,140), (8, 24,240), (10, 21,670), (10, 24,1640), (11, 12,80), (14, 20,930), (14, 22,760), (15, 16,520), (15, 17,700), (16, 17,280), (16, 18,330), (17, 24,270), (19, 20,130), (19, 21,420), (20, 23,250), (21, 23,300), (22, 23,550)]
#Definieer de coordinaten van de knopen in de graaf
coord = {0:(107, 26),1:(226.75, 103),2:(182.25, 167.25),3:(97, 178),4:(80, 100), 5:(67.25, 125), 
6:(112.25, 112.25), 7:(122.25, 147.75), 8:(172.25, 82.25), 9:(200, 182.75), 10:(217.75, 40), 
11:(112.58, 144.85), 12:(115, 138.42), 13:(138.42, 156.58), 14:(137, 129), 15:(116.58, 92.4), 
16:(125.42, 65.16), 17:(146.54, 71.85), 18:(101.5, 58.77), 19:(180.17, 86.98), 20:(176.46, 98.67), 
21:(202.56, 90.21), 22:(176.58, 141.98), 23:(186.42, 109),24:(161.46,78.70)}
#Definieer de graaf G
G = nx.Graph()
#Voeg alle takken toe aan de graaf
G.add_weighted_edges_from(edges)
```

Teken deze graaf met de knopen op de juiste locaties gedefinieerd in de lijst <i>coord</i>.

```python
#[Antwoord]
#Teken de graaf G met de knopen op de juiste locatie
nx.draw_networkx(G,pos=coord, node_size=5, with_labels=False)
```

Zoals we weten is voor dit probleem in networkx geen algoritme gedefinieerd, omdat het voor een graaf met veel knopen veel te arbeidsintensief is om alle oplossingen door te rekenen en er geen algoritme bestaat dat gegarandeerd op een snelle manier de optimale oplossing vindt.     
Hier willen we echter alleen de knopen 0 t/m 10 aandoen en dat is nog net niet teveel om alle oplossingen door te rekenen.     
We gaan echter alleen van de wegen uit en er is geen rechtstreekse verbinding tussen ieder paar knopen. M.b.v. Dijkstra's algoritme kunnen we echter wel voor ieder paar knopen in de graaf de minimale afstand door dit wegennetwerk bepalen. 

## Vraag b.    
Bepaal voor ieder paar knopen in de graaf G de minimale afstand over het wegennetwerk weergegeven in graaf G.

```python
afstand=dict(nx.all_pairs_dijkstra_path_length(G)) 
print(afstand[0][9])
```

Nu we alle afstanden hebben berekend, kunnen we van ieder mogelijk circuit langs alle knopen 0 t/m 10 de totale lengte berekenen om op die manier het kortste circuit te vinden.     
## Vraag c.     
Maak gebruik van de code zoals beschikbaar bij het dictaat H2.2. Pas deze code waar nodig aan met de afstanden die je bij vraag b. hebt gevonden om op deze manier het optimale circuit te vinden. Ga er vanuit dat je fietsroute start in knoop 0.

```python
#[Antwoord]

#Importeer de procedure permutations uit de module itertools
from itertools import permutations
from math import inf

#Zet initiële waarde van de totale lengte van het circuit op oneindig
min_lengte = inf
#Doorloop alle permutaties van de lijst [1,..,10]
for perm in list(permutations(range(1,11))):
#Zet het startpunt 0 voor en achter de permutatie om een volledig circuit te definieren
    circuit = [0]+list(perm)+[0]
    
#Bereken de lengte van het totale circuit 
#HIER MAAK JE GEBRUIK VAN DE EERDER BEREKENDE AFSTANDEN
    lengte = sum(afstand[circuit[i]][circuit[i+1]] for i in range(len(circuit)-1))
#Als de lengte kleiner is dan tot nu toe
    if lengte < min_lengte:
#Onthoud de nieuwe minimale lengte en het beste circuit 
        min_lengte = lengte
        best_circuit = circuit

print("De beste route is:", best_circuit, 'ter lengte', min_lengte, 'meter')
        
```

## Vraag d.     
Teken de route die je hebt gevonden in een graaf. Houd er rekening mee dat hiervoor extra knopen nodig zijn, omdat sommige kortste routes tussen twee knopen indirect langs andere knopen kunnen lopen. 

```python
circuit=[]
for i in range(len(best_circuit)-1):
    pad=nx.dijkstra_path(G,best_circuit[i], best_circuit[i+1])
    circuit+=pad
print(circuit)
G1=nx.path_graph(circuit)
lussen=[(i,j) for i,j in G1.edges() if i==j]
G1.remove_edges_from(lussen)
nx.draw_networkx(G1,coord)
plt.show()

```

```python
print("Totale lengte:", sum(afstand[circuit[i]][circuit[i+1]] for i in range(len(circuit)-1)))
```
