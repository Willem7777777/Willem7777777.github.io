De kamers zijn de punten 0 t/m 4 in graaf G. De belangrijke locaties zijn de punten 5 t/m 10. De overige punten zijn gewoon kruispunten in de graaf. In de cel hieronder wordt deze graaf G gedefinieerd. Run deze cel door in het menu hierboven Cell - Run All te kiezen

```python
#Importeer de benodigde modules
import networkx as nx
from matplotlib import pyplot as plt

#Definieer de takken van de graaf met hun lengte als gewicht
edges = [(0, 16,900),(0, 18,630),(1, 9,1630), (1, 21,500), (2, 9,300), (2, 13,800), (2, 22,590), (3, 5,1110), (3, 11,580), (3, 13,920), (4, 5,380), (4, 6,670), (4, 18,750), (6, 12,640), (6, 14,990), (6, 15,600), (7, 11,100), (7, 12,110), (7, 13,260), (7, 14,410), (8, 19,140), (8, 24,240), (10, 21,670), (10, 24,1640), (11, 12,80), (14, 20,930), (14, 22,760), (15, 16,520), (15, 17,700), (16, 17,280), (16, 18,330), (17, 24,270), (19, 20,130), (19, 21,420), (20, 23,250), (21, 23,300), (22, 23,550)]
#Definieer de coordinaten van de knopen in de graaf
coord = [[107, 26], [226.75, 103], [182.25, 167.25], [97, 178], [80, 100], [67.25, 125], 
[112.25, 112.25], [122.25, 147.75], [172.25, 82.25], [200, 182.75], [217.75, 40], 
[112.58, 144.85], [115, 138.42], [138.42, 156.58], [137, 129], [116.58, 92.4], 
[125.42, 65.16], [146.54, 71.85], [101.5, 58.77], [180.17, 86.98], [176.46, 98.67], 
[202.56, 90.21], [176.58, 141.98], [186.42, 109],[161.46,78.70]]
#Definieer de graaf G
G = nx.Graph()
#Voeg alle takken toe aan de graaf
G.add_weighted_edges_from(edges)

#Teken de graaf G met de knopen op de juiste locatie
nx.draw_networkx(G,coord)
plt.show()
```
