```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
doelpunten = np.array([12,13,60,3,2,21,0,11,8])
print(doelpunten)
```

```python
df = pd.DataFrame(data=doelpunten, columns=['doelpunten'])
df
```

```python
df_nls = df.drop([2])
df_nls
```

```python
mean_dp = doelpunten.mean()
std_dp = doelpunten.std(ddof=1)

mean_dp_nls = df_nls['doelpunten'].mean()
std_dp_nls = df_nls['doelpunten'].std()
print(mean_dp)
```

```python
val = 0. # this is the value where you want the data to appear on the y-axis.
plt.plot(doelpunten, np.zeros_like(doelpunten) + val, 'o', 
         mean_dp, np.zeros_like(mean_dp) + val, 'rx',
        np.array([mean_dp-std_dp,mean_dp+std_dp]),
         np.zeros_like(np.array([mean_dp-std_dp,mean_dp+std_dp])),'r|', 
         markersize=14, markeredgewidth=2)
plt.yticks([])
plt.ylim(-4, 4)
plt.axes().set_aspect('equal')
plt.title('Aantal doelpunten aanvallers Ajax 2009/2010')
plt.show()
```

```python
val = 0. # this is the value where you want the data to appear on the y-axis.
plt.plot(doelpunten, np.zeros_like(doelpunten) + val, 'o', 
         mean_dp, np.zeros_like(mean_dp) + val, 'rx',
        np.array([mean_dp-std_dp,mean_dp+std_dp]),
         np.zeros_like(np.array([mean_dp-std_dp,mean_dp+std_dp])),'r|',
        mean_dp_nls, np.zeros_like(mean_dp_nls) + val, 'kx',
        np.array([mean_dp_nls-std_dp_nls,mean_dp_nls+std_dp_nls]),
         np.zeros_like(np.array([mean_dp_nls-std_dp_nls,mean_dp_nls+std_dp_nls])),'k|', 
         markersize=14, markeredgewidth=2)
plt.yticks([])
plt.ylim(-4, 4)
plt.axes().set_aspect('equal')
plt.title('Aantal doelpunten aanvallers Ajax 2009/2010')
plt.show()
```

```python
median_dp = np.median(doelpunten)
q75, q25 = np.percentile(doelpunten, [75 ,25])
q75
```
