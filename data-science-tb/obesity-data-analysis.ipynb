```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# Lees het CSV-bestand in een DataFrame

# Definieer de onafhankelijke variabele (X) en afhankelijke variabele (y)
X = df[['Age']]
y = df['NObeyesdad']

# Maak een instantie van het logistische regressiemodel
model = LogisticRegression()

# Pas het model aan de gegevens aan
model.fit(X, y)

# Voorspel de kans op obesitas voor nieuwe gegevens
nieuwe_gegevens = pd.DataFrame({'Age': [30, 40, 50]})
kansen = model.predict_proba(nieuwe_gegevens)[:, 1]

print("Kansen op insuffiecient weight:", kansen)




```

```python
import numpy as np
from scipy.stats import chi2_contingency

# Voorbeeldgegevens
onafhankelijke_variabele = df['family_history_with_overweight']
afhankelijke_variabele = df['Insufficient_Weight']

# Maak een kruistabel
kruistabel = np.array([onafhankelijke_variabele, afhankelijke_variabele])

# Voer de Chi-kwadraattoets uit
chi2, p_value, dof, expected = chi2_contingency(kruistabel)

# Resultaten weergeven
print("Chi-kwadraatstatistiek:", chi2)
print("P-waarde:", p_value)
print("Vrijheidsgraden:", dof)
print("Verwachte frequenties:")
print(expected)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Voorbeeldgegevens

# Definieer numerieke waarden voor NObeyesdad
nobeysdad_mapping = {
    'Insufficient Weight': 0,
    'Normal Weight': 1,
    'Overweight Level I': 2,
    'Overweight Level II': 3,
    'Obesity Type I': 4,
    'Obesity Type II': 5,
    'Obesity Type III': 6
}

df['NObeyesdad_numeric'] = df['NObeyesdad'].map(nobeysdad_mapping)

# Maak het histogram
plt.hist2d(df['FAVC'], df['NObeyesdad_numeric'], cmap='Blues')

# Labels en titel toevoegen
plt.xlabel('FAVC')
plt.ylabel('NObeyesdad')
plt.title('Histogram van FAVC en NObeyesdad')

# Kleurenbalk toevoegen
plt.colorbar()

# Grafiek weergeven
plt.show()
```
